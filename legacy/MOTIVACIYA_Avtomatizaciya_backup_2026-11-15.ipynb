```python
import os
from IPython.display import Audio, display
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GOOGLE_API_KEY"))

PROJECT = "/content/drive/MyDrive/YouTube_Auto/Long_Videos"
audio_folder = os.path.join(PROJECT, "audio")
os.makedirs(audio_folder, exist_ok=True)

text = """
Ты можешь изменить свою жизнь.
Но сначала тебе нужно перестать откладывать важные решения.
Настоящая сила начинается тогда, когда ты продолжаешь действовать, даже если тебе трудно.
"""

voices = ["Orus", "Charon", "Algieba"]

for voice in voices:
    print("Генерация голоса:", voice)

    result = client.interactions.create(
        model="gemini-3.8-flash-tts",
        input=[{
            "type": "user_input",
            "content": [{
                "type": "text",
                "text": text,
                "annotations": [{
                    "type": "speech_metadata",
                    "style": "Deep, low-pitched mature male narrator. Calm, confident, slightly gravelly, natural Russian pronunciation. Documentary narration, emotionally expressive but not shouting. Speak slowly with meaningful pauses."
                }]
            }]
        }],
        response_format={"type": "audio"},
        generation_config={
            "speech_config": [{"voice": voice}]
        }
    )

    path = os.path.join(audio_folder, f"test_{voice}.wav")

    with open(path, "wb") as f:
        f.write(result.output_audio.data)

    print("Даяр:", voice)
    display(Audio(path))
```

```python
import os
import base64
from IPython.display import Audio, display
from google import genai
from google.colab import userdata

client = genai.Client(api_key=userdata.get("GOOGLE_API_KEY"))

PROJECT = "/content/drive/MyDrive/YouTube_Auto/Long_Videos"
audio_folder = os.path.join(PROJECT, "audio")
os.makedirs(audio_folder, exist_ok=True)

text = """
Ты можешь изменить свою жизнь.
Но сначала тебе нужно перестать откладывать важные решения.
Настоящая сила начинается тогда, когда ты продолжаешь действовать, даже если тебе трудно.
"""

voices = ["Orus", "Charon", "Algieba"]

for voice in voices:
    print("Генерация голоса:", voice)

    result = client.interactions.create(
        model="gemini-3.8-flash-tts",
        input=[{
            "type": "user_input",
            "content": [{
                "type": "text",
                "text": text,
                "annotations": [{
                    "type": "speech_metadata",
                    "style": "Deep, low-pitched mature male narrator. Calm, confident, slightly gravelly, natural Russian pronunciation. Documentary narration, emotionally expressive but not shouting. Speak slowly with meaningful pauses."
                }]
            }]
        }],
        response_format={"type": "audio"},
        generation_config={"speech_config": [{"voice": voice}]}
    )

    audio_data = base64.b64decode(result.output_audio.data)
    path = os.path.join(audio_folder, f"test_{voice}.wav")

    with open(path, "wb") as f:
        f.write(audio_data)

    print("Даяр:", voice)
    display(Audio(audio_data))
```

```python
import os
import re
import base64
import subprocess

# Биринчи бөлүктүн чек арасын сактоо
first_chunk = script[:1800].rsplit(" ", 1)[0]
remaining = script[len(first_chunk):].strip()

# Сүйлөмдөр боюнча бөлүү
sentences = re.split(r'(?<=[.!?…])\s+', remaining)

chunks = []
current = ""

for sentence in sentences:
    if not sentence.strip():
        continue
    if len(current) + len(sentence) > 1500 and current:
        chunks.append(current.strip())
        current = sentence
    else:
        current += " " + sentence

if current.strip():
    chunks.append(current.strip())

print("Калган бөлүктөр:", len(chunks))

# Ар бир бөлүктү озвучкалоо
style = """
Deep, warm, mature Russian male voice, Orus.
Calm, sincere and emotionally expressive motivational storytelling.
Natural pauses, varied intonation, clear Russian pronunciation.
Do not rush. Avoid monotonous delivery.
"""

for i, chunk_text in enumerate(chunks, start=2):
    raw_path = os.path.join(audio_folder, f"video_001_part_{i:02d}_raw.wav")
    slow_path = os.path.join(audio_folder, f"video_001_part_{i:02d}_slow.wav")

    if os.path.exists(slow_path):
        print(f"{i}-бөлүк мурда даярдалган, өткөрүлдү.")
        continue

    print(f"{i}-бөлүк даярдалууда...")

    result = client.interactions.create(
        model="gemini-3.8-flash-tts",
        input=style + "\n\n" + chunk_text,
        response_format={"type": "audio"},
        generation_config={
            "speech_config": [{"voice": "Orus"}]
        }
    )

    audio_data = base64.b64decode(result.output_audio.data)

    with open(raw_path, "wb") as f:
        f.write(audio_data)

    subprocess.run([
        "ffmpeg", "-y", "-i", raw_path,
        "-filter:a", "atempo=0.85",
        slow_path
    ], check=True, capture_output=True)

    print(f"{i}-бөлүк сакталды!")

print("Калган аудиолорду даярдоо аяктады!")
```

```python
import os
import base64
import subprocess
from google import genai
from google.colab import userdata

# 1. Кардарды жана ачкычты кайра жүктөө
key = userdata.get("GOOGLE_API_KEY")
client = genai.Client(api_key=key)

audio_folder = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/audio"
scripts_folder = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/scripts"
os.makedirs(audio_folder, exist_ok=True)

# 2. Сценарийди окуп, абзацтарга бөлүү (chunks тизмесин калыбына келтирүү)
script_path = os.path.join(scripts_folder, "video_001_ready.txt")
if os.path.exists(script_path):
    with open(script_path, "r", encoding="utf-8") as f:
        script_text = f.read()

    paragraphs = [p.strip() for p in script_text.split("\n") if p.strip()]

    # Текстти топторго бөлүү алгоритми
    from itertools import combinations
    n = len(paragraphs)
    target = sum(map(len, paragraphs)) / 6
    best = None
    best_score = float("inf")
    for cuts in combinations(range(1, n), 5):
        bounds = (0,) + cuts + (n,)
        groups = [
            "\n\n".join(paragraphs[bounds[i]:bounds[i+1]])
            for i in range(6)
        ]
        score = sum((len(g) - target) ** 2 for g in groups)
        if score < best_score:
            best_score = score
            best = groups
    chunks = best
else:
    raise FileNotFoundError(f"Сценарий файлы табылган жок: {script_path}")

# 3. Аудио генерация стили
style = """
Deep, warm, mature Russian male voice.
Calm, sincere and emotionally expressive motivational storytelling.
Natural pauses, varied intonation.
Do not rush. Avoid monotonous delivery.
"""

print("1-бөлүк даярдалууда...")

result = client.interactions.create(
    model="gemini-3.8-flash-tts",
    input=style + "\n\n" + chunks[0],
    response_format={"type": "audio"},
    generation_config={
        "speech_config": [{"voice": "Orus"}]
    }
)

audio_data = base64.b64decode(result.output_audio.data)

raw_path = audio_folder + "/video_001_part_01_same6_raw.wav"
slow_path = audio_folder + "/video_001_part_01_same6_slow.wav"

with open(raw_path, "wb") as f:
    f.write(audio_data)

subprocess.run([
    "ffmpeg", "-y", "-i", raw_path,
    "-filter:a", "atempo=0.85",
    slow_path
], check=True, capture_output=True)

print("1-БӨЛҮК ДАЯР!")
print("Файл:", slow_path)
```

```python
topics = [
    "man working laptop",
    "man thinking alone",
    "alarm clock waking up",
    "stressed man office",
    "writing goals notebook",
    "man walking alone",
    "sunrise nature"
]

downloaded = 0
seen_ids = {3135}

for topic in topics:
    params = {
        "key": pixabay_key,
        "q": topic,
        "per_page": 5,
        "safesearch": "true"
    }

    data = requests.get(
        "https://pixabay.com/api/videos/",
        params=params,
        timeout=30
    ).json()

    for item in data.get("hits", []):
        if item["id"] in seen_ids:
            continue

        files = item.get("videos", {})
        quality = "large" if "large" in files else "medium"
        video_url = files[quality]["url"]

        downloaded += 1
        path = os.path.join(
            clips_folder,
            f"clip_{downloaded+1:02d}.mp4"
        )

        try:
            response = requests.get(video_url, timeout=120)
            response.raise_for_status()

            with open(path, "wb") as f:
                f.write(response.content)

            seen_ids.add(item["id"])
            print("Жүктөлдү:", topic, "ID:", item["id"])

        except Exception as e:
            print("Ката:", topic, e)

        if downloaded >= 14:
            break

    if downloaded >= 14:
        break

print("Жалпы жаңы жүктөлгөн видео:", downloaded)
```

```python
import os
import requests
import time

save_dir = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/clips_2k"
os.makedirs(save_dir, exist_ok=True)

queries = [
    "man thinking",
    "man using phone",
    "alarm clock",
    "stressed man",
    "writing checklist",
    "man walking sunrise"
]

used_ids = set()
downloaded = []

for i, q in enumerate(queries, 1):
    print(f"\n{i}/6. Изделүүдө: {q}")

    params = {
        "key": PIXABAY_API_KEY,
        "q": q,
        "per_page": 20,
        "safesearch": "true"
    }

    data = requests.get(
        "https://pixabay.com/api/videos/",
        params=params,
        timeout=30
    ).json()

    candidates = [
        v for v in data.get("hits", [])
        if v["id"] not in used_ids
        and v.get("videos")
    ]

    candidates.sort(
        key=lambda v: max(
            (x.get("width", 0) * x.get("height", 0)
             for x in v["videos"].values()),
            default=0
        ),
        reverse=True
    )

    if not candidates:
        print("Бул тема боюнча видео табылган жок.")
        continue

    video = candidates[0]
    sizes = video["videos"]
    best = max(
        sizes.values(),
        key=lambda x: x.get("width", 0) * x.get("height", 0)
    )

    path = os.path.join(save_dir, f"clip_{i:02d}.mp4")

    print("Жүктөлүүдө:", video.get("tags"))
    print("Өлчөмү:", best.get("width"), "x", best.get("height"))

    r = requests.get(best["url"], timeout=120)
    r.raise_for_status()

    with open(path, "wb") as f:
        f.write(r.content)

    used_ids.add(video["id"])
    downloaded.append(path)
    print("Даяр:", path)

print("\nЖалпы жүктөлгөн видео:", len(downloaded))
```

```python
import os
import requests

save_dir = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/clips_2k"
os.makedirs(save_dir, exist_ok=True)

queries = [
    ("procrastination", "new_16.mp4"),
    ("writing goals notebook", "new_17.mp4")
]

for query, filename in queries:
    print("Изделүүдө:", query)

    params = {
        "key": PIXABAY_API_KEY,
        "q": query,
        "per_page": 30,
        "safesearch": "true"
    }

    data = requests.get(
        "https://pixabay.com/api/videos/",
        params=params,
        timeout=30
    ).json()

    hits = [
        v for v in data.get("hits", [])
        if v.get("duration", 0) >= 5
        and v.get("videos")
    ]

    hits.sort(
        key=lambda v: max(
            x.get("width", 0) * x.get("height", 0)
            for x in v["videos"].values()
        ),
        reverse=True
    )

    if not hits:
        print("Видео табылган жок:", query)
        continue

    v = hits[0]
    sizes = v["videos"]
    landscape = [
        x for x in sizes.values()
        if x.get("width", 0) >= x.get("height", 0)
    ]
    best = max(
        landscape or list(sizes.values()),
        key=lambda x: x.get("width", 0) * x.get("height", 0)
    )

    response = requests.get(best["url"], timeout=180)
    response.raise_for_status()

    path = os.path.join(save_dir, filename)
    with open(path, "wb") as f:
        f.write(response.content)

    print("Даяр:", filename, best["width"], "x", best["height"])

print("Эки видео боюнча издөө аяктады.")
```

```python
import os, requests, getpass
from PIL import Image
from io import BytesIO
from IPython.display import display

# API ачкычы мурда сакталбаса, кайра сурайт
if "PIXABAY_API_KEY" not in globals():
    PIXABAY_API_KEY = getpass.getpass("Pixabay API ачкычыңды киргиз: ")

folder = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/clips_2k"
os.makedirs(folder, exist_ok=True)

queries = [
    "man sitting at desk thinking",
    "calendar deadline planning",
    "writing goals notebook",
    "man starting work laptop",
    "focused man working",
    "man walking sunrise city"
]

downloaded = []
used_urls = set()

for i, q in enumerate(queries, 1):
    print(f"{i}/6 Изделүүдө: {q}")

    try:
        r = requests.get(
            "https://pixabay.com/api/videos/",
            params={
                "key": PIXABAY_API_KEY,
                "q": q,
                "per_page": 10,
                "video_type": "film",
                "safesearch": "true"
            },
            timeout=30
        )
        r.raise_for_status()
        hits = r.json().get("hits", [])

        # Горизонталдуу, сапаттуу видеону тандайбыз
        candidates = []
        for v in hits:
            files = v.get("videos", {})
            f = files.get("large") or files.get("medium")
            if f and f.get("width", 0) >= 1280 and f.get("width", 0) > f.get("height", 0):
                candidates.append(f)

        candidates.sort(key=lambda x: x.get("width", 0), reverse=True)
        chosen = next((f for f in candidates if f["url"] not in used_urls), None)

        if not chosen:
            print("  Бул темада ылайыктуу кадр табылган жок.")
            continue

        url = chosen["url"]
        used_urls.add(url)

        path = os.path.join(folder, f"extra_{i:02d}.mp4")
        data = requests.get(url, timeout=90)
        data.raise_for_status()

        with open(path, "wb") as file:
            file.write(data.content)

        downloaded.append(path)
        print(f"  Даяр: extra_{i:02d}.mp4")

    except Exception as e:
        print("  Ката:", str(e)[:180])

print("\nЖүктөлгөн видеолор:", len(downloaded))
print("Сакталган папка:", folder)
```

```python

import os, re, subprocess

root = "/content/drive/MyDrive/YouTube_Auto/Long_Videos"
video = root + "/test_2k/ANIMATED_60SEC_MODERN.mp4"
srt = root + "/subs/subtitles_60sec.srt"
ass = root + "/subs/subtitles_dynamic.ass"
output = root + "/test_2k/DYNAMIC_SUBTITLES_60SEC.mp4"

# Файлдарды текшерүү
for f in [video, srt]:
    if not os.path.exists(f):
        raise FileNotFoundError("Файл табылган жок: " + f)

def to_seconds(t):
    h, m, rest = t.split(":")
    sec, ms = rest.split(",")
    return int(h)*3600 + int(m)*60 + int(sec) + int(ms)/1000

# SRT окуу
with open(srt, "r", encoding="utf-8-sig") as f:
    content = f.read()

blocks = re.split(r"\n\s*\n", content.strip())

def ass_time(seconds):
    h = int(seconds // 3600)
    m = int((seconds % 3600) // 60)
    s = seconds % 60
    return f"{h}:{m:02d}:{s:05.2f}"

lines = [
    "[Script Info]",
    "ScriptType: v4.00+",
    "PlayResX: 2560",
    "PlayResY: 1440",
    "ScaledBorderAndShadow: yes",
    "",
    "[V4+ Styles]",
    "Format: Name, Fontname, Fontsize, PrimaryColour, SecondaryColour, OutlineColour, BackColour, Bold, Italic, BorderStyle, Outline, Shadow, Alignment, MarginL, MarginR, MarginV, Encoding",
    "Style: Dynamic,Arial,76,&H00FFFFFF,&H0000FFFF,&H00000000,&H80000000,-1,0,1,6,3,2,100,100,125,1",
    "",
    "[Events]",
    "Format: Layer, Start, End, Style, Name, MarginL, MarginR, MarginV, Effect, Text"
]

for block in blocks:
    rows = block.strip().splitlines()
    if len(rows) < 3 or "-->" not in rows[1]:
        continue

    start_text, end_text = [x.strip() for x in rows[1].split("-->")]
    start = to_seconds(start_text)
    end = to_seconds(end_text)
    text = " ".join(rows[2:]).strip()
    words = re.findall(r"\S+", text)

    if not words:
        continue

    duration_cs = max(1, round((end-start)*100))
    weights = [max(1, len(re.sub(r"[^\w]", "", w))) for w in words]
    total = sum(weights)
    parts = [max(1, round(duration_cs*w/total)) for w in weights]
    parts[-1] += duration_cs - sum(parts)

    karaoke = " ".join(
        "{\\kf" + str(max(1, p)) + "}" + w.replace("{", "").replace("}", "")
        for w, p in zip(words, parts)
    )

    lines.append(
        f"Dialogue: 0,{ass_time(start)},{ass_time(end)},Dynamic,,0,0,0,,{karaoke}"
    )

with open(ass, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

print("Динамик субтитр даяр!")
print("Эми видео монтаждалат. Бир нече мүнөт күт.")

cmd = [
    "ffmpeg", "-y", "-i", video,
    "-vf", f"ass={ass}",
    "-c:v", "libx264", "-preset", "ultrafast",
    "-crf", "20", "-c:a", "copy",
    output
]

result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    print("ДАЯР! Жаңы видео:", output)
else:
    print(result.stderr[-4000:])
```

```python

import os
import subprocess

base = "/content/drive/MyDrive/YouTube_Auto/Long_Videos"
clips_folder = base + "/clips_2k"

audio = "/content/test_audio_60sec.wav"
subs = base + "/subs/subtitles_60sec_modern.ass"
output = base + "/test_2k/cinematic_test_60sec.mp4"

selected_clips = [
    "clip_02.mp4", "clip_03.mp4", "clip_04.mp4",
    "extra_02.mp4", "extra_03.mp4", "extra_04.mp4",
    "extra_05.mp4", "new_01.mp4", "new_02.mp4",
    "new_05.mp4", "new_06.mp4", "new_07.mp4",
    "new_12.mp4", "new_14.mp4", "new_15.mp4"
]

assert os.path.exists(audio), "Аудио табылган жок"
assert os.path.exists(subs), "Субтитр табылган жок"
assert all(os.path.exists(os.path.join(clips_folder, f))
           for f in selected_clips), "Кайсы бир видео табылган жок"

cmd = ["ffmpeg", "-y"]

for name in selected_clips:
    cmd += ["-i", os.path.join(clips_folder, name)]

cmd += ["-i", audio]

filters = []

for i in range(15):
    filters.append(
        f"[{i}:v]fps=30,trim=duration=4,setpts=PTS-STARTPTS,"
        f"scale=2560:1440:force_original_aspect_ratio=increase,"
        f"crop=2560:1440,"
        f"zoompan=z='min(zoom+0.0007,1.05)':"
        f"x='iw/2-(iw/zoom/2)':y='ih/2-(ih/zoom/2)':"
        f"d=1:s=2560x1440:fps=30,setsar=1[v{i}]"
    )

filters.append(
    "".join(f"[v{i}]" for i in range(15)) +
    "concat=n=15:v=1:a=0[v]"
)

filters.append(
    f"[v]subtitles='{subs}':"
    "force_style='FontName=DejaVu Sans,FontSize=44,"
    "PrimaryColour=&H0000FFFF,OutlineColour=&H00000000,"
    "BorderStyle=1,Outline=3,Shadow=1'[outv]"
)

cmd += [
    "-filter_complex", ";".join(filters),
    "-map", "[outv]",
    "-map", "15:a",
    "-t", "60",
    "-c:v", "libx264",
    "-preset", "veryfast",
    "-crf", "20",
    "-pix_fmt", "yuv420p",
    "-c:a", "aac",
    "-b:a", "192k",
    "-shortest",
    output
]

print("🎬 Cinematic монтаж башталды...")
result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    print("✅ ДАЯР:", output)
else:
    print("❌ КАТА:")
    print(result.stderr[-4000:])
```

```python
import os

base = "/content/drive/MyDrive/YouTube_Auto/Long_Videos/subs"
source = os.path.join(base, "subtitles_60sec_modern.ass")
target = os.path.join(base, "subtitles_60sec_cinematic.ass")

with open(source, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

updated = False

for i, line in enumerate(lines):
    if line.startswith("Style:"):
        parts = line.rstrip("\n").split(",")
        parts[2] = "68"             # Чоң шрифт
        parts[3] = "&H0000FFFF"     # Сары түс
        parts[5] = "&H00000000"     # Кара контур
        parts[7] = "-1"             # Калың
        parts[15] = "1"             # Контур стили
        parts[16] = "5"             # Контур калыңдыгы
        parts[17] = "2"             # Көлөкө
        parts[18] = "2"             # Төмөндө жайгашуу
        parts[21] = "100"           # Төмөнкү аралык
        lines[i] = ",".join(parts) + "\n"
        updated = True

with open(target, "w", encoding="utf-8-sig") as f:
    f.writelines(lines)

print("✅ Жаңы чоң субтитр даяр!")
print(target)
print("Үн менен дал келүү убакыттары сакталды.")
```

```python
import os

base = "/content/drive/MyDrive/YouTube_Auto/Long_Videos"
source = os.path.join(base, "subs", "subtitles_60sec_modern.ass")
output = os.path.join(base, "subs", "subtitles_60sec_cinematic.ass")

with open(source, "r", encoding="utf-8-sig") as f:
    lines = f.readlines()

for i, line in enumerate(lines):
    if line.startswith("Style: Default,"):
        parts = line.rstrip("\n").split(",")
        parts[1] = "DejaVu Sans"
        parts[2] = "68"
        parts[7] = "-1"
        parts[15] = "1"
        parts[16] = "5"
        parts[17] = "3"
        parts[18] = "2"
        parts[21] = "100"
        lines[i] = ",".join(parts) + "\n"

for i, line in enumerate(lines):
    if line.startswith("Dialogue:"):
        parts = line.rstrip("\n").split(",", 9)
        if len(parts) == 10:
            parts[9] = (
                r"{\fad(180,120)\fscx85\fscy85"
                r"\t(0,220,\fscx108\fscy108)"
                r"\t(220,320,\fscx100\fscy100)}"
                + parts[9]
            )
            lines[i] = ",".join(parts) + "\n"

with open(output, "w", encoding="utf-8-sig") as f:
    f.writelines(lines)

print("✅ Чоң кинематографиялык субтитр даяр!")
print(output)
```
