Ар бир видео өзүнчө папкада: `MyDrive/YouTube_Auto/Videos/<дата>_<тема>/`


```python
#@title ▶️ ВИДЕО ЖАСОО — теманы жазып, ушул баскычты басыңыз
TOPIC = "Почему ты откладываешь жизнь на завтра: психология прокрастинации"  #@param {type:"string"}
NICHE = "motivation"  #@param ["motivation"]
MODE = "Автомат (жаңы же улантуу)"  #@param ["Автомат (жаңы же улантуу)", "Жаңы папка (ушул тема менен кайра)", "Этаптан кайра иштетүү"]
RERUN_STAGE = "s1_script"  #@param ["s1_script", "s2_scenes", "s3_voice", "s4_subtitles", "s5_visuals", "s6_music", "s7_render", "s8_qa"]
STOP_AFTER = "Аягына чейин"  #@param ["Аягына чейин", "s1_script", "s2_scenes", "s3_voice", "s4_subtitles", "s5_visuals", "s6_music", "s7_render"]

# ---------------- Төмөнкүнү өзгөртпөңүз ----------------
REPO = "https://github.com/Nazigul338/YouTube_Auto_Pipeline.git"
BRANCH = "claude/youtube-auto-pipeline-87d41t"
CODE_DIR = "/content/YouTube_Auto_Pipeline"

import os, sys, subprocess, importlib

def sh(cmd):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if r.returncode != 0:
        raise RuntimeError(f"Буйрук иштеген жок: {cmd}\n{r.stderr[-2000:]}")
    return r.stdout

# 1) Google Drive
from google.colab import drive
if not os.path.exists("/content/drive/MyDrive"):
    drive.mount("/content/drive")
print("✓ Google Drive туташты")

# 2) Коддун акыркы версиясы GitHub'дан
if os.path.exists(f"{CODE_DIR}/.git"):
    sh(f"git -C {CODE_DIR} fetch -q origin {BRANCH} && git -C {CODE_DIR} reset -q --hard FETCH_HEAD")
else:
    sh(f"git clone -q -b {BRANCH} {REPO} {CODE_DIR}")
print("✓ Код жаңыланды:", sh(f"git -C {CODE_DIR} log -1 --format='%h %s'").strip())

# 3) Китепканалар (бир жолу гана орнотулат)
try:
    import google.genai, yaml, requests  # noqa
except ImportError:
    sh(f"pip install -q -r {CODE_DIR}/requirements.txt")
print("✓ Китепканалар даяр\n")

# 4) Иштетүү
if CODE_DIR not in sys.path:
    sys.path.insert(0, CODE_DIR)
for m in [m for m in list(sys.modules) if m == "pipeline" or m.startswith("pipeline.")]:
    del sys.modules[m]          # жаңыланган кодду кайра жүктөө
from pipeline.runner import run

run(
    TOPIC,
    niche=NICHE,
    new=MODE.startswith("Жаңы"),
    rerun_from=RERUN_STAGE if MODE.startswith("Этаптан") else None,
    stop_after=None if STOP_AFTER == "Аягына чейин" else STOP_AFTER,
)
```
